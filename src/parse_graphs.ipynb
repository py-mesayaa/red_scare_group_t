```python
path_to_graphs = "../red-scare/data/"
filename = "G-ex.txt"
filename_dir = "increase-n10-1.txt"

with open(path_to_graphs + filename_dir, 'r') as f:
    lines = [line.strip() for line in f if line.strip()]

# Line 0: n_vertices, n_edges, n_red
n, m, r = map(int, lines[0].split())

# Line 1: start_vertex, end_vertex
s, t = map(int, lines[1].split())

red_vertices = set()

# The vertices are located from index 2 up to 2 + n
adj_list = {}
for i in range(2, 2 + n):
    parts = lines[i].split()
    vertex_id = int(parts[0])
    adj_list[vertex_id] = []
    # If there is a second part and it is a '*', the vertex is red
    if len(parts) > 1 and parts[1] == '*':
        red_vertices.add(vertex_id)

for i in range(2+n, len(lines)):
    v_str, sym, u_str = lines[i].split(" ")
    v, u = int(v_str), int(u_str)
    if sym == "--":
        adj_list[v].append(u)
        adj_list[u].append(v)
    if sym == "->":
        adj_list[v].append(u)
adj_list
```
