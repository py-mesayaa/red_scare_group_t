```python
from build_graphs import to_algs4, parse_graph
from itu.algs4.graphs.breadth_first_paths import BreadthFirstPaths
from pathlib import Path


filename = "G-ex"

path = f"../red-scare/data/{filename}.txt"
```

```python
adj_list, red_vertices, s, t, directed = parse_graph(path=path)
```

```python
def none(adj_list, red_vertices, s, t, directed):
    def is_valid(x):
        # check if vertex x is valid (not red or is s or t)
        return x not in red_vertices or x in (s, t)
    
    G, index, names = to_algs4(adj_list, directed, keep_edge = lambda v, u: is_valid(v) and is_valid(u))

    bfs = BreadthFirstPaths(G, index[s])

    if not bfs.has_path_to(index[t]):
        return -1
    return bfs.dist_to(index[t])
```

```python
result = none(adj_list, red_vertices, s, t, directed)
result
```

```python
out = Path("../red-scare/output")
out.mkdir(parents=True, exist_ok=True)
(out / "none_answers.txt").write_text(f"{filename} : {result}\n")
```
